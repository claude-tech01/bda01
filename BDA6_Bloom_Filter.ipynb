```python
class BloomFilter:

    def __init__(self, size):
        self.size = size
        self.bit_array = [0] * size

    # First hash function
    def hash1(self, item):
        return hash(item) % self.size

    # Second hash function
    def hash2(self, item):
        return (hash(item) * 7 + 13) % self.size

    # Insert an element
    def add(self, item):
        index1 = self.hash1(item)
        index2 = self.hash2(item)

        self.bit_array[index1] = 1
        self.bit_array[index2] = 1

    # Check whether an element exists
    def check(self, item):
        index1 = self.hash1(item)
        index2 = self.hash2(item)

        if (self.bit_array[index1] == 1 and
            self.bit_array[index2] == 1):
            return "Possibly Present"
        else:
            return "Definitely Not Present"


# Main Program
bloom = BloomFilter(20)

# Insert elements
elements = ["apple", "banana", "mango", "orange"]

for item in elements:
    bloom.add(item)

print("Inserted Elements:", elements)

print("\nBloom Filter Bit Array:")
print(bloom.bit_array)

# Search elements
print("\nSearching Elements:")

print("Apple:", bloom.check("apple"))
print("Banana:", bloom.check("banana"))
print("Grapes:", bloom.check("grapes"))
print("Pineapple:", bloom.check("pineapple"))
```
