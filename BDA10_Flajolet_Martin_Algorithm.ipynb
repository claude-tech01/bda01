```python

# Experiment 10: Flajolet-Martin Algorithm

import hashlib

# Step 1: Hash function
def hash_value(item):
    hash_result = hashlib.sha256(
        str(item).encode()
    ).hexdigest()

    return int(hash_result, 16)


# Step 2: Count trailing zeros
def count_trailing_zeros(number):
    if number == 0:
        return 256

    binary = bin(number)[2:]
    return len(binary) - len(binary.rstrip('0'))


# Step 3: Flajolet-Martin Algorithm
def flajolet_martin(data):

    max_zeros = 0

    for item in data:

        hashed = hash_value(item)

        zeros = count_trailing_zeros(hashed)

        max_zeros = max(max_zeros, zeros)

    # Step 4: Estimate distinct elements
    estimate = 2 ** max_zeros

    return estimate


# Step 5: Input data stream
data_stream = [
    10, 20, 30, 40, 10,
    20, 50, 60, 30, 70,
    80, 90, 100, 50, 110
]

# Step 6: Calculate result
actual_count = len(set(data_stream))

estimated_count = flajolet_martin(data_stream)

print("Input Data Stream:")
print(data_stream)

print("\nActual Distinct Elements:")
print(actual_count)

print("\nEstimated Distinct Elements:")
print(estimated_count)
```
