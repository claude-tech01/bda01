```python

# Experiment 11: MapReduce using Python
# Operations: Joins, Sorting and Searching


# ==========================================
# PART 1: MAPREDUCE JOIN
# ==========================================

print("========== MAPREDUCE JOIN ==========")

# Student dataset
students = [
    (1, "Rahul"),
    (2, "Amit"),
    (3, "Sneha"),
    (4, "Priya")
]

# Marks dataset
marks = [
    (1, 85),
    (2, 90),
    (3, 78),
    (5, 88)
]

# MAP PHASE
mapped = []

for sid, name in students:
    mapped.append((sid, ("Student", name)))

for sid, mark in marks:
    mapped.append((sid, ("Marks", mark)))

# SHUFFLE AND SORT PHASE
mapped.sort(key=lambda x: x[0])

# REDUCE PHASE
grouped = {}

for sid, value in mapped:
    grouped.setdefault(sid, []).append(value)

print("Joined Student Records:")

for sid, records in grouped.items():
    student_name = None
    student_marks = None

    for record_type, value in records:
        if record_type == "Student":
            student_name = value
        else:
            student_marks = value

    if student_name is not None and student_marks is not None:
        print(sid, student_name, student_marks)


# ==========================================
# PART 2: MAPREDUCE SORTING
# ==========================================

print("\n========== MAPREDUCE SORTING ==========")

numbers = [45, 12, 78, 34, 23, 89, 10, 56]

# MAP PHASE
mapped_numbers = [(number, 1) for number in numbers]

# SHUFFLE AND SORT PHASE
mapped_numbers.sort(key=lambda x: x[0])

# REDUCE PHASE
sorted_numbers = [number for number, count in mapped_numbers]

print("Original Numbers:", numbers)
print("Sorted Numbers:", sorted_numbers)


# ==========================================
# PART 3: MAPREDUCE SEARCHING
# ==========================================

print("\n========== MAPREDUCE SEARCHING ==========")

data = [10, 20, 30, 40, 50, 60, 70, 80]

search_element = 50

# MAP PHASE
mapped_data = []

for number in data:
    if number == search_element:
        mapped_data.append((number, "Found"))

# REDUCE PHASE
if mapped_data:
    print(search_element, "is Found in the Dataset")
else:
    print(search_element, "is Not Found in the Dataset")
```
