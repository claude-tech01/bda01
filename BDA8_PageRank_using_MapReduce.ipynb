```python

# Experiment 8: PageRank Algorithm using MapReduce

# Step 1: Define the web graph
graph = {
    'A': ['B', 'C'],
    'B': ['C'],
    'C': ['A'],
    'D': ['C']
}

# Step 2: Initialize PageRank values
N = len(graph)
d = 0.85

pagerank = {page: 1/N for page in graph}

# Step 3: MapReduce iterations
iterations = 10

for iteration in range(iterations):

    # MAP PHASE
    mapped = []

    for page, links in graph.items():
        if links:
            rank_share = pagerank[page] / len(links)

            for link in links:
                mapped.append((link, rank_share))

    # REDUCE PHASE
    new_rank = {page: 0 for page in graph}

    for page, rank in mapped:
        new_rank[page] += rank

    # Apply PageRank formula
    for page in graph:
        new_rank[page] = (
            (1 - d) / N + d * new_rank[page]
        )

    pagerank = new_rank

    print(f"Iteration {iteration + 1}:")
    print(pagerank)

# Step 4: Display final PageRank
print("\nFinal PageRank Scores:")

for page, rank in sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(page, ":", round(rank, 4))

print("\nHighest Ranked Page:",
      max(pagerank, key=pagerank.get))
```
